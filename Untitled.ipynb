```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('parrot (1).jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
averaging = cv2.blur(img, (5, 5))

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(averaging, cv2.COLOR_BGR2RGB))
plt.title("Averaging Filter")
plt.axis('off')
plt.show()
```

```python
kernel = np.array([[1, 2, 1],
                   [2, 4, 2],
                   [1, 2, 1]], dtype=np.float32) / 16

weighted = cv2.filter2D(img, -1, kernel)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB))
plt.title("Weighted Averaging Filter")
plt.axis('off')
plt.show()
```

```python
gaussian = cv2.GaussianBlur(img, (5, 5), 0)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Filter")
plt.axis('off')
plt.show()
```

```python
median = cv2.medianBlur(img, 5)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title("Median Filter")
plt.axis('off')
plt.show()
```

```python
laplacian_kernel = np.array([[0, -1, 0],
                             [-1, 5, -1],
                             [0, -1, 0]])

sharpened = cv2.filter2D(img, -1, laplacian_kernel)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
plt.title("Laplacian Sharpening Using Kernel")
plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(6, 6))
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Operator")
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(3, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')

plt.subplot(3, 3, 2)
plt.imshow(cv2.cvtColor(averaging, cv2.COLOR_BGR2RGB))
plt.title("Averaging Filter")
plt.axis('off')

plt.subplot(3, 3, 3)
plt.imshow(cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB))
plt.title("Weighted Averaging")
plt.axis('off')

plt.subplot(3, 3, 4)
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Filter")
plt.axis('off')

plt.subplot(3, 3, 5)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title("Median Filter")
plt.axis('off')

plt.subplot(3, 3, 6)
plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
plt.title("Laplacian Sharpening")
plt.axis('off')

plt.subplot(3, 3, 7)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Operator")
plt.axis('off')

plt.tight_layout()
plt.show()
```
